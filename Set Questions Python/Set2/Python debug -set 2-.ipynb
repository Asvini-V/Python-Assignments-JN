```python
#2.Find the error in the following Python code:
def my_function():
    x += 5
    print(x)

x = 10
my_function()
```

```python
# SOLUTION: x is a global variable, but inside the function it is treated as local
#           So we must declare x as global inside the function before modifying it

def my_function():
    global x
    x += 5
    print(x)

x = 10
my_function()
```

```python
# SOLUTION: list.sort() sorts the list in-place and returns None
#           So printing numbers.sort() will print None
#           Instead, first sort the list, then print the list

numbers = [5, 3, 8, 2]
numbers.sort()
print(numbers)
```

```python
#5.
def square(n):
    n * n

print(square(4))
```

```python
# SOLUTION: The function is not returning the result
#           We must use 'return' to send back the value

def square(n):
    return n * n

print(square(4))
```

```python
# SOLUTION: Two format specifiers %s and %d are used
#           But only one variable (name) is given
#           We must pass both name and age as a tuple

name = "Alice"
age = 25

print("My name is %s and I am %d years old" % (name, age))
```

```python
# SOLUTION: 'is' checks whether both variables refer to the same object (memory location)
#           But here a and b are two different list objects with same values
#           So we must use '==' to compare values

a = [1, 2, 3]
b = [1, 2, 3]

if a == b:
    print("Lists are equal")
```
